```python
import numpy as np
import pandas as pd
import warnings
warnings.simplefilter("ignore", UserWarning)
```

```python
df = pd.read_csv('./Dataset/iris.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
enc = LabelEncoder()
```

```python
df['species'] = enc.fit_transform(df['species'])
```

```python
df.head()
```

```python
df = df[df['species'] != 0] [['sepal_width','petal_length','species']]
```

```python
df.head()
```

```python
df = df.sample(100)
df_train = df.iloc[:60,:].sample(10) #random 10 training data
df_val = df.iloc[60:80,:].sample(5) #random 5 validation data
df_test = df.iloc[80:,:].sample(5) #random 5 test data
```

```python
X_test = df_val.iloc[:,0:2].values
y_test = df_val.iloc[:,-1].values
```

```python
df_bagging = df_train.sample(7, replace = True)
X = df_bagging.iloc[:,0:2]
y = df_bagging.iloc[:,-1]
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from mlxtend.plotting import plot_decision_regions
from sklearn.metrics import accuracy_score
```

```python
dt_bagging1 = DecisionTreeClassifier()
```

```python
dt_bagging1.fit(X, y)
y_pred = dt_bagging1.predict(X_test)
print(accuracy_score(y_test, y_pred))
```

```python
models = []
accuracies = []

for i in range(7):

    # Bootstrap sample
    sample = df_train.sample(7, replace=True)

    X_train = sample.iloc[:,0:2]
    y_train = sample.iloc[:,-1]

    model = DecisionTreeClassifier(random_state=i)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)

    print(f"Tree {i+1} Accuracy = {acc:.2f}")

    models.append(model)
    accuracies.append(acc)
```

```python
query = np.array([2.2, 5.0]).reshape(1, 2)

for i, model in enumerate(models):
    print(f"Tree {i+1}:", model.predict(query))
```
