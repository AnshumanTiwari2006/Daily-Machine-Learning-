```python
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python
#convertors
pd.read_csv('./Dataset/matches.csv', converters = {'team2': rename}).head(3)
```
