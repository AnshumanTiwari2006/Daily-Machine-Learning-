### Load the Titanic Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Titanic dataset
titanic_df = sns.load_dataset('titanic')

display(titanic_df.head())
```

### Dataset Information and Descriptive Statistics

```python
# Display general information about the DataFrame
titanic_df.info()
```

```python
# Display descriptive statistics for numerical columns
display(titanic_df.describe())
```

### Missing Values Analysis

```python
# Check for missing values
display(titanic_df.isnull().sum())
```

### Distribution of Numerical Features

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.histplot(titanic_df['age'].dropna(), kde=True)
plt.title('Distribution of Age')

plt.subplot(1, 2, 2)
sns.histplot(titanic_df['fare'], kde=True)
plt.title('Distribution of Fare')

plt.tight_layout()
plt.show()
```

### Distribution of Categorical Features

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.countplot(x='survived', data=titanic_df)
plt.title('Survival Count')

plt.subplot(1, 3, 2)
sns.countplot(x='sex', data=titanic_df)
plt.title('Gender Count')

plt.subplot(1, 3, 3)
sns.countplot(x='pclass', data=titanic_df)
plt.title('Passenger Class Count')

plt.tight_layout()
plt.show()
```

### Survival Rate by Gender and Passenger Class

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
sns.barplot(x='sex', y='survived', data=titanic_df)
plt.title('Survival Rate by Gender')

plt.subplot(1, 2, 2)
sns.barplot(x='pclass', y='survived', data=titanic_df)
plt.title('Survival Rate by Passenger Class')

plt.tight_layout()
plt.show()
```

### Age Distribution by Survival Status

```python
plt.figure(figsize=(8, 6))
sns.histplot(x='age', hue='survived', data=titanic_df, kde=True, multiple='stack')
plt.title('Age Distribution by Survival Status')
plt.show()
```

### Correlation Heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(titanic_df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```
