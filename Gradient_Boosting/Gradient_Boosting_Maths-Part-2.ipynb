```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor
```

```python
X = np.array([
    [1, 10, 100],
    [2, 20, 200],
    [3, 30, 300]
])

y = np.array([10, 20, 30])
```

```python
M = 2                  
learning_rate = 1.0 
```

```python
# STEP 1:
# f0(x) = argmin_gamma Sum L(yi, gamma)
#
# For squared error:
# f0 = mean(y)
```

```python
f = np.full(len(y), np.mean(y))

print("Initial prediction f0:")
print(f)
```

```python
# Gradient boosting
```

```python
trees = []
gammas = []


for m in range(M):
    print(f"BOOSTING ITERATION {m + 1}")
```

```python
# STEP 2(a)
    #
    # r_im = - dL/df
    #
    # For squared error:
    #
    # L = 1/2 (y - f)^2
    #
    # Negative gradient = y - f
```

```python
residual = y - f

print("\nPseudo-residuals:")
print(residual)
```

```python
    # STEP 2(b)
    #
    # Fit regression tree to pseudo-residuals
```

```python
tree = DecisionTreeRegressor(max_depth=1, random_state=42)
tree.fit(X, residual)
    # Tree prediction = approximate pseudo-residual
tree_prediction = tree.predict(X)
print("\nTree prediction:")
print(tree_prediction)
```

```python
    # STEP 2(c)
    #
    # Calculate gamma for each leaf.
    #
    # For squared error:
    #
    # gamma = mean residual in the leaf
```

```python
    leaf_ids = tree.apply(X)

    unique_leaves = np.unique(leaf_ids)

    gamma_dict = {}

    print("\nLeaf calculations:")

    for leaf in unique_leaves:

        indices = np.where(leaf_ids == leaf)[0]

        leaf_residuals = residual[indices]

        gamma = np.mean(leaf_residuals)

        gamma_dict[leaf] = gamma

        print(f"Leaf {leaf}")
        print("Rows:", indices)
        print("Residuals:", leaf_residuals)
        print("Gamma:", gamma)
```

```python
  # STEP 2(d)
    #
    # f_m(x)
    #
    # = f_(m-1)(x)
    # + learning_rate * gamma
```

```python
    correction = np.array([
        gamma_dict[leaf]
        for leaf in leaf_ids
    ])

    f = f + learning_rate * correction

    print("\nCorrection:")
    print(correction)

    print("\nUpdated prediction:")
    print(f)


    trees.append(tree)
    gammas.append(gamma_dict)
```

```python
print("FINAL PREDICTIONS")
print(f)
```
